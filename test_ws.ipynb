```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# URL للبحث عن وظائف Data Scientist
url = 'https://wuzzuf.net/search/jobs?q=data+scientist'

# إرسال طلب HTTP
headers = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.content, 'html.parser')

# قوائم لتخزين البيانات
job_titles = []
company_names = []
locations = []
job_times = []
required_skills = []

# استخراج البيانات من كل بطاقة وظيفة
jobs = soup.find_all('div', class_='css-1gatmva e1v1l3u10')  # قد يتغير class مع الوقت

for job in jobs:
    # اسم الوظيفة
    title = job.find('h2', class_='css-m604qf').text.strip()
    job_titles.append(title)
    
    # اسم الشركة
    company = job.find('a', class_='css-17s97q8').text.strip()
    company_names.append(company)
    
    # الموقع
    location = job.find('span', class_='css-5wys0k').text.strip()
    locations.append(location)
    
    # نوع العمل (دوام كامل/جزئي)
    job_time = job.find('span', class_='css-1ve4b75 eoyjyou0').text.strip()
    job_times.append(job_time)
    
    # المهارات المطلوبة
    skills = job.find('div', class_='css-y4udm8').text.strip()
    required_skills.append(skills)

# إنشاء DataFrame
jobs_df = pd.DataFrame({
    'Job Title': job_titles,
    'Company': company_names,
    'Location': locations,
    'Job Type': job_times,
    'Required Skills': required_skills
})
print(jobs_df.head())  # هيعرض أول 5 وظائف في الجدول
# حفظ البيانات في ملف CSV
jobs_df.to_csv('wuzzuf_jobs.csv', index=False)
```
